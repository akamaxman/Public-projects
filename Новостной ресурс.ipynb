```python
# Пишем функцию нормализации дат
def normalize_date(date_str):
    if pd.isna(date_str): # Проверяем пропущенные значения
        return pd.NA # Возвращаем специальное значение для пропущенных данных

    parts = str(date_str).split('.')  # Делим строку даты по точкам
    parts_count = len(parts)  # Считаем количество полученных часте

    # Обрабатываем день
    day = parts[0].zfill(2) if parts_count >= 1 else '01' # Добавляем ноль если нужно

    # Обрабатываем месяц
    month = parts[1].zfill(2) if parts_count >= 2 else '01' # Аналогично добавляем ноль

    # Обрабатываем год
    if parts_count >= 3: # Если год указан
        year = parts[2] if len(parts[2]) == 4 else '1900' # Проверяем формат года
    else: # Если год не указан
        year = '1900' # Устанавливаем год по умолчанию

    # Формируем нормализованную дату
    normalized = f"{day}.{month}.{year}"

    # Проверяем валидность даты(существует ли такая дата)
    try:
        pd.to_datetime(normalized, format='%d.%m.%Y') # Пробуем преобразовать в datetime
        return normalized
    except:
        return pd.NA # Если дата невалидна - возвращаем пропущенное значение
```

### Решение V2 при помощи NumPy(работает намного быстрее!)

```python
import numpy as np

# Создаем копию исходного DataFrame для отчета
original_dates = dates.copy()

# Функция для нормализации дат с использованием только pandas
def normalize_dates_pandas(df):
    # Разделяем дату на компоненты
    split_date = df['date'].str.split('.', expand=True)

    # Заполняем компоненты даты
    day = split_date[0].fillna('01').str.zfill(2)
    month = split_date[1].fillna('01').str.zfill(2)
    year = split_date[2].fillna('1900')

    # Исправляем годы с неправильным форматом
    year = np.where(year.str.len() != 4, '1900', year)

    # Собираем дату обратно
    normalized = day + '.' + month + '.' + year

    # Проверяем валидность даты
    valid_mask = pd.to_datetime(normalized, format='%d.%m.%Y', errors='coerce').notna()
    df['normalized_date'] = np.where(valid_mask, normalized, np.nan)

    return df

# Применяем нормализацию
dates = normalize_dates_pandas(dates)

# Заменяем исходный столбец и удаляем временный
dates['date'] = dates['normalized_date']
dates.drop(columns=['normalized_date'], inplace=True)

# Создаем отчет
report = pd.DataFrame({
    'original': original_dates['date'],
    'normalized': dates['date']
})

# Вычисляем статистику
stats = {
    'total_records': len(dates),
    'corrected_records': (report['original'] != report['normalized']).sum(),
    'null_values_before': report['original'].isna().sum(),
    'null_values_after': report['normalized'].isna().sum(),
    'examples': report.head(10).to_dict('records')
}

# Выводим отчет
print("Отчет о нормализации дат:")
print(f"Всего записей: {stats['total_records']}")
print(f"Исправлено записей: {stats['corrected_records']}")
print(f"Пустых значений до обработки: {stats['null_values_before']}")
print(f"Пустых значений после обработки: {stats['null_values_after']}")
print("\nПримеры преобразований:")
for example in stats['examples']:
    print(f"{example['original']} -> {example['normalized']}")
```

```python
# Импорт необходимых библиотек
import feedparser  # Для парсинга RSS-лент
import requests  # Для работы с HTTP-запросами (API новостей)
import pandas as pd  # Для работы с табличными данными
from datetime import datetime  # Для работы с датами
import matplotlib.pyplot as plt  # Для визуализации данных
import json  # Для работы с JSON-форматом
import time  # Для работы с временными задержками
from collections import defaultdict  # Для удобной работы со словарями
import os  # Для работы с файловой системой
```

```python
# Конфиг программы в виде словаря
CONFIG = {
    # Создаём список политических субъектов для мониторинга
    "political_entities": ["Путин", "Зеленский", "Байден", "Трамп", "Единая Россия", "КПРФ", "США", "Россия", "Украина"],
    # Создаём список RSS-лент для анализа
    "rss_feeds": [
        "https://lenta.ru/rss/news",
        "https://www.vedomosti.ru/rss/news",
        "https://www.kommersant.ru/RSS/news.xml"
    ],
    # Настройки для NewsAPI
    "newsapi_config": {
        "api_key": "a61bf49cc7f9437a93190a3f0c3b5cba",  # Ключ API
        "sources": "bbc-news,cnn,reuters",  # Источники новостей
        "language": "ru"  # Язык новостей
    },
    # Настройки хранения данных
    "storage": {
        "raw_data": "data/raw_news.json",  # Путь к файлу с сырыми данными
        "processed_data": "data/processed_news.csv",  # Путь к обработанным данным
        "plots": "plots/"  # Директория для графиков
    },
    # Интервал обновления данных в секундах
    "update_interval": 3600
}

class NewsCollector:
    """
    Класс для сбора новостей из RSS-лент и по API.
    Основные функции:
    - fetch_rss_news: сбор новостей из RSS-лент
    - fetch_newsapi_news: сбор новостей через NewsAPI
    - collect_news: основной метод сбора данных
    - save_raw_data: сохранение сырых данных
    """

    def __init__(self, config):
        """Инициализация коллектора с конфигурацией"""
        self.config = config  # Сохраняем конфигурацию
        self.ensure_data_directories()  # Создаем необходимые директории

    def ensure_data_directories(self):
        """Создает необходимые директории для хранения данных, если они не существуют"""
        # os.makedirs создает директории, exist_ok=True предотвращает ошибку если директория уже существует
        os.makedirs(os.path.dirname(self.config["storage"]["raw_data"]), exist_ok=True)
        os.makedirs(self.config["storage"]["plots"], exist_ok=True)

    def fetch_rss_news(self):
        """
        Сбор новостей из RSS-лент.
        Возвращает список новостей, содержащих упоминания политических субъектов.
        """
        news_items = []  # Пкстой список для хранения найденных новостей

        # Перебираем все RSS-ленты из конфигурации
        for rss_url in self.config["rss_feeds"]:
            try:
                # Парсим RSS-ленту
                feed = feedparser.parse(rss_url)

                # Перебираем все записи в RSS-ленте
                for entry in feed.entries:
                    # Создаем словарь с информацией о новости
                    news_item = {
                        "title": entry.get("title", ""),  # Заголовок новости
                        "description": entry.get("description", ""),  # Описание
                        "link": entry.get("link", ""),  # Ссылка на новость
                        "published": entry.get("published", ""),  # Дата публикации
                        "source": "RSS: " + rss_url,  # Источник новости
                        "contains_entities": []  # Список упомянутых политических субъектов
                    }

                    # Проверяем упоминания каждого политического субъекта
                    for entity in self.config["political_entities"]:
                        # Ищем упоминание в заголовке или описании (без учета регистра)
                        if entity.lower() in news_item["title"].lower() or entity.lower() in news_item["description"].lower():
                            news_item["contains_entities"].append(entity)  # Добавляем в список

                    # Если есть упоминания, добавляем новость в результат
                    if news_item["contains_entities"]:
                        news_items.append(news_item)

            except Exception as e:
                # В случае ошибки выводим сообщение и продолжаем работу
                print(f"Ошибка при обработке RSS {rss_url}: {e}")

        return news_items

    def fetch_newsapi_news(self):
        """
        Сбор новостей через NewsAPI.
        Для работы требуется API ключ от newsapi.org.
        """
        news_items = []  # Пустой список для хранения новостей
        api_key = self.config["newsapi_config"]["api_key"]  # Получаем API ключ

        # Проверяем, установлен ли API ключ
        if api_key == "YOUR_API_KEY":
            print("Предупреждение: NewsAPI не будет использоваться, так как не установлен API ключ")
            return news_items  # Возвращаем пустой список если ключ не установлен

        base_url = "https://newsapi.org/v2/everything"  # Базовый URL API

        # Перебираем все политические субъекты
        for entity in self.config["political_entities"]:
            try:
                # Параметры запроса
                params = {
                    "q": entity,  # Поисковый запрос
                    "sources": self.config["newsapi_config"]["sources"],  # Источники
                    "language": self.config["newsapi_config"]["language"],  # Язык
                    "apiKey": api_key  # API ключ
                }

                # Отправляем get-запрос к API
                response = requests.get(base_url, params=params)
                response.raise_for_status()  # Проверяем на ошибки
                data = response.json()  # Парсим JSON-ответ

                # Обрабатываем полученные новости
                for article in data.get("articles", []):
                    # Создаем словарь с информацией о новости
                    news_item = {
                        "title": article.get("title", ""), # Заголовок новости
                        "description": article.get("description", ""), # Описание
                        "link": article.get("url", ""), # Ссылка на новость
                        "published": article.get("publishedAt", ""), # Дата
                        "source": "NewsAPI: " + article.get("source", {}).get("name", ""), # Ичточник
                        "contains_entities": [entity]  # Записываем найденный субъект
                    }
                    news_items.append(news_item)  # Добавляем новость в список

                # Делаем паузу между запросами для соблюдения лимитов NewsAPI
                time.sleep(1)

            except Exception as e:
                # В случае ошибки выводим сообщение и продолжаем работу
                print(f"Ошибка при запросе NewsAPI для '{entity}': {e}")

        return news_items

    def collect_news(self):
        """
        Основной метод сбора новостей из всех источников.
        Объединяет результаты из RSS и API.
        """
        print("Начало сбора новостей...")
        # Получаем новости из RSS
        rss_news = self.fetch_rss_news()
        # Получаем новости из NewsAPI
        api_news = self.fetch_newsapi_news()
        # Объединяем результаты
        all_news = rss_news + api_news

        # Выводим статистику
        print(f"Собрано {len(all_news)} новостей (RSS: {len(rss_news)}, API: {len(api_news)})")
        return all_news

    def save_raw_data(self, news_data):
        """
        Сохраняет сырые данные в JSON файл.
        """
        # Открываем файл для записи
        with open(self.config["storage"]["raw_data"], "w", encoding="utf-8") as f:
            # Записываем данные в JSON с отступами
            json.dump(news_data, f, ensure_ascii=False, indent=2)
        print(f"Сырые данные сохранены в {self.config['storage']['raw_data']}")

class NewsAnalyzer:
    """
    Класс для анализа собранных новостей.
    Основные функции:
    - process_news_data: обработка сырых данных
    - analyze_mentions: анализ частоты упоминаний
    - visualize_results: визуализация результатов
    """

    def __init__(self, config):
        """Инициализация анализатора с конфигурацией"""
        self.config = config  # Сохраняем конфигурацию

    def load_raw_data(self):
        """
        Загружает сырые данные из файла.
        Возвращает список новостей или пустой список при ошибке.
        """
        try:
            # Открываем файл для чтения
            with open(self.config["storage"]["raw_data"], "r", encoding="utf-8") as f:
                return json.load(f)  # Загружаем JSON данные
        except FileNotFoundError:
            print("Файл с сырыми данными не найден")
            return []  # Возвращаем пустой список если файл не найден

    def process_news_data(self, news_data):
        """
        Обрабатывает новостные данные и создает dataframe для анализа.
        Разбирает даты, нормализует структуру данных.
        """
        processed_data = []  # Пустой список для обработанных данных

        # Перебираем все новости
        for news_item in news_data:
            try:
                # Получаем дату публикации
                pub_date = news_item.get("published", "")
                try:
                    # Пробуем разобрать дату в формате ISO (NewsAPI)
                    pub_date = datetime.strptime(pub_date, "%Y-%m-%dT%H:%M:%SZ")
                except:
                    try:
                        # Пробуем разобрать дату в формате RSS
                        pub_date = datetime.strptime(pub_date, "%a, %d %b %Y %H:%M:%S %z")
                    except:
                        # Если не удалось, используем текущую дату
                        pub_date = datetime.now()

                # Для каждого упомянутого субъекта создаем отдельную запись
                for entity in news_item["contains_entities"]:
                    processed_data.append({
                        "date": pub_date.date(),  # Дата без времени
                        "entity": entity,  # Политический субъект
                        "title": news_item["title"],  # Заголовок новости
                        "source": news_item["source"],  # Источник
                        "link": news_item["link"]  # Ссылка
                    })

            except Exception as e:
                # В случае ошибки выводим сообщение и продолжаем
                print(f"Ошибка при обработке новости: {e}")
                continue

        # Создаем dataframe из обработанных данных
        df = pd.DataFrame(processed_data)
        return df

    def analyze_mentions(self, df):
        """
        Анализирует частоту упоминаний по дням и источникам.
        Возвращает два dataframe:
        - daily_mentions: упоминания по дням
        - source_mentions: упоминания по источникам
        """
        if df.empty:
            return None, None  # Возвращаем None если данных нет

        # Группируем по дате и субъекту, считаем количество
        daily_mentions = df.groupby(["date", "entity"]).size().unstack(fill_value=0)

        # Группируем по источнику и субъекту, считаем количество
        source_mentions = df.groupby(["source", "entity"]).size().unstack(fill_value=0)

        return daily_mentions, source_mentions

    def save_processed_data(self, df):
        """
        Сохраняет обработанные данные в CSV файл.
        """
        # Сохраняем dataframe в CSV без индекса
        df.to_csv(self.config["storage"]["processed_data"], index=False, encoding="utf-8")
        print(f"Обработанные данные сохранены в {self.config['storage']['processed_data']}")

    def visualize_results(self, daily_mentions, source_mentions):
        """
        Создает визуализации результатов анализа.
        Генерирует два графика:
        1. Динамика упоминаний по дням
        2. Распределение упоминаний по источникам
        """
        # Проверяем, есть ли данные для визуализации
        if daily_mentions is None or source_mentions is None:
            print("Нет данных для визуализации")
            return

        # Создаем график динамики упоминаний
        plt.figure(figsize=(12, 6))  # Задаем размер графика
        # Рисуем линии для каждого политического субъекта
        for column in daily_mentions.columns:
            plt.plot(daily_mentions.index, daily_mentions[column], label=column)

        # Настраиваем внешний вид графика
        plt.title("Динамика упоминаний политических субъектов")
        plt.xlabel("Дата")
        plt.ylabel("Количество упоминаний")
        plt.legend()  # Добавляем легенду
        plt.grid()  # Добавляем сетку
        plt.xticks(rotation=45)  # Поворачиваем подписи дат
        plt.tight_layout()  # Оптимизируем расположение элементов

        # Сохраняем график в файл
        daily_plot_path = os.path.join(self.config["storage"]["plots"], "daily_mentions.png")
        plt.savefig(daily_plot_path)
        print(f"График динамики сохранен в {daily_plot_path}")
        plt.close()  # Закрываем график для освобождения памяти

        # Создаем график распределения по источникам
        source_mentions.plot(kind="bar", stacked=True, figsize=(12, 6))
        plt.title("Упоминания политических субъектов по источникам")
        plt.xlabel("Источник")
        plt.ylabel("Количество упоминаний")
        plt.xticks(rotation=45)
        plt.tight_layout()

        # Сохраняем второй график
        source_plot_path = os.path.join(self.config["storage"]["plots"], "source_mentions.png")
        plt.savefig(source_plot_path)
        print(f"График по источникам сохранен в {source_plot_path}")
        plt.close() # Закрываем график для освобождения памяти

def main():
    """
    Основная функция программы.
    Организует бесконечный цикл сбора и анализа данных.
    """
    # Инициализация классов
    collector = NewsCollector(CONFIG)  # Создаем сборщик новостей
    analyzer = NewsAnalyzer(CONFIG)  # Создаем анализатор

    try:
        # Бесконечный цикл работы программы
        while True:
            # 1. Сбор данных
            news_data = collector.collect_news()  # Собираем новости
            collector.save_raw_data(news_data)  # Сохраняем сырые данные

            # 2. Анализ данных
            df = analyzer.process_news_data(news_data)  # Обрабатываем данные
            if not df.empty:  # Если есть данные
                analyzer.save_processed_data(df)  # Сохраняем обработанные данные
                # Анализируем упоминания
                daily_mentions, source_mentions = analyzer.analyze_mentions(df)
                # Визуализируем результаты
                analyzer.visualize_results(daily_mentions, source_mentions)

            # 3. Ожидание следующего обновления
            print(f"Ожидание следующего обновления через {CONFIG['update_interval']} секунд...")
            time.sleep(CONFIG["update_interval"])  # Пауза

    except KeyboardInterrupt:
        # Обработка прерывания программы (Ctrl+C)
        print("Программа остановлена пользователем")

if __name__ == "__main__":
    # Точка входа в программу
    main()
```
